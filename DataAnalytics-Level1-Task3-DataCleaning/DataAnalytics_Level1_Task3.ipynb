```python
import pandas as pd
import numpy as np

# Exact 'Housing.csv' file load karein
df = pd.read_csv('Housing.csv')

# Initial inspection
print("--- Initial Data Shape ---")
print(df.shape)

print("\n--- Missing Values Count ---")
print(df.isnull().sum())

print("\n--- Duplicate Rows Count ---")
print(df.duplicated().sum())

df.head()
```

```python
# Numeric columns ko Median se aur Text columns ko Mode se fill karein
for col in df.columns:
    if df[col].dtype in ['float64', 'int64']:
        df[col] = df[col].fillna(df[col].median())
    else:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values after imputation:")
print(df.isnull().sum())
```

```python
# Duplicates remove karein
df = df.drop_duplicates()

# Categorical/Text data ko clean karein (lowercase aur extra spaces remove)
text_columns = df.select_dtypes(include=['object']).columns
for col in text_columns:
    df[col] = df[col].astype(str).str.strip().str.lower()

print("Text columns cleaned:", list(text_columns))
```

```python
# Numerical columns ke Outliers ko IQR method se cap karein
num_columns = df.select_dtypes(include=[np.number]).columns
for col in num_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    df[col] = np.where(df[col] < lower_limit, lower_limit, df[col])
    df[col] = np.where(df[col] > upper_limit, upper_limit, df[col])

# Cleaned data ko CSV file mein save karein
df.to_csv('cleaned_housing.csv', index=False)
print("Data Cleaning Complete! File saved as 'cleaned_housing.csv'")
```
